Simulation of the Manna model and saving of the `train_pool` and `test_pool` sets to the `data_repository/grids/` directory. One run per system size (environment variable `SOC_L`). `MAX_POS` must exceed the number of extremes in a run. If the number of collected positives equals the limit, the limit has to be raised.

---

```python
import gc
import os
import sys

sys.path.insert(0, "../data_management")
import M1_Manna as m

L = int(os.environ.get("SOC_L", 32))
N = int(os.environ.get("SOC_N", 500000000))
SEED = int(os.environ.get("SOC_SEED", 12345))
MAX_POS = {32: 12000, 64: 5000, 96: 3500, 128: 2500}.get(L, 5000)
MAX_NEG = int(os.environ.get("SOC_MAX_NEG", 10000))
```

#### Warm-up

```python
grid, _, _, _, _, _ = m.run_for("load", L=L, seed=SEED)
```

#### Training phase

```python
grid, s, q, g, lab, idx = m.run_for("train", L=L, N=N, grid=grid, max_pos=MAX_POS, max_neg=MAX_NEG, seed=SEED)
print(f"positives {int(lab.sum())} (limit {MAX_POS})")
m.save_phase("train_pool", L, grid=grid, s_all=s, q_all=q, grids=g, labels=lab, indices=idx)
del s, q, g, lab, idx
gc.collect()
```

#### Test phase 

```python
grid, s, q, g, lab, idx = m.run_for("test", L=L, N=N, grid=grid, max_pos=MAX_POS, max_neg=MAX_NEG, seed=SEED + 1)
print(f"positives {int(lab.sum())} (limit {MAX_POS})")
m.save_phase("test_pool", L, grid=grid, s_all=s, q_all=q, grids=g, labels=lab, indices=idx)
del grid, s, q, g, lab, idx
gc.collect()
```
